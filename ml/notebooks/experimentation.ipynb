```python
import sys
from pathlib import Path

# Project-er ml folder Python path-e add kora
possible_paths = [
    Path.cwd() / "ml",
    Path.cwd().parent,
    Path.cwd().parent.parent / "ml"
]

for path in possible_paths:
    if (path / "skill_mapping").exists():
        sys.path.insert(0, str(path))
        break

from skill_mapping.skill_gap import calculate_skill_gap
from skill_mapping.skill_scoring import calculate_skill_score
from matching.similarity import calculate_similarity
from matching.skill_matcher import match_opportunities


# Student-er existing skills
student_skills = [
    "Python",
    "SQL",
    "Machine Learning"
]

# Required skills
required_skills = [
    "Python",
    "SQL",
    "Machine Learning",
    "Deep Learning",
    "NLP"
]


# 1. Skill Gap Analysis
gap_result = calculate_skill_gap(
    student_skills,
    required_skills
)

print("===== SKILL GAP ANALYSIS =====")
print("Matched Skills:", gap_result["matched_skills"])
print("Missing Skills:", gap_result["missing_skills"])
print("Matching Percentage:",
      str(gap_result["matching_percentage"]) + "%")


# 2. Skill Scoring
skill_scores = {
    "Python": 85,
    "SQL": 70,
    "Machine Learning": 60
}

score_result = calculate_skill_score(skill_scores)

print("\n===== SKILL SCORING =====")
print("Skill Scores:", score_result["skill_scores"])
print("Overall Score:",
      score_result["overall_score"])


# 3. Similarity Calculation
similarity_result = calculate_similarity(
    student_skills,
    required_skills
)

print("\n===== SKILL SIMILARITY =====")
print("Similarity Score:",
      similarity_result["similarity_score"])
print("Similarity Percentage:",
      str(similarity_result["similarity_percentage"]) + "%")


# 4. Opportunity Matching
opportunities = [
    {
        "name": "Machine Learning Internship",
        "required_skills": [
            "Python",
            "Machine Learning",
            "SQL"
        ]
    },
    {
        "name": "Data Science Internship",
        "required_skills": [
            "Python",
            "SQL",
            "Machine Learning",
            "Deep Learning"
        ]
    },
    {
        "name": "Web Development Internship",
        "required_skills": [
            "HTML",
            "CSS",
            "JavaScript"
        ]
    }
]

matching_result = match_opportunities(
    student_skills,
    opportunities
)

print("\n===== OPPORTUNITY MATCHING =====")

for opportunity in matching_result:
    print(
        opportunity["name"],
        "| Match Score:",
        str(opportunity["match_score"]) + "%"
    )
```
